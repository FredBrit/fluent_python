```python
# вертикальное объединение двух np.array

a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)

res=np.concatenate((a,b), axis=0)

# Или
# np.vstack((a, b))

# Или 
# np.r_[a, b]
res
```

```python
# горизонтальное объединение двух np.array

a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)

res=np.concatenate((a,b), axis=1)
res
```

```python
# Генерация повторяющейся последовательности элементов по разным паттернам

a = np.array([1,2,3])

res=np.repeat(a, 3)
print(res)

res2=np.tile(a,3)
print(res2)

result=np.r_[np.repeat(a, 3), np.tile(a,3)]
result
```

```python
# Поиск общих уникальных элементов в двух массивах

a = np.array([1,2,3,2,3,4,3,4,5,6])
b = np.array([7,2,10,2,7,4,9,4,9,8])

res=np.intersect1d(a,b)
res
```

```python
# Фильтрация массива по наличию общих элементов с другим массивом

a = np.array([1,2,3,4,5])
b = np.array([5,6,7,8,9])


res=np.setdiff1d(a,b)
res
```

```python
# Поиск позиций, где элементы совпадают

a = np.array([1,2,3,2,3,4,3,4,5,6])
b = np.array([7,2,10,2,7,4,9,4,9,8])

res=np.where(a==b)
res
```

```python
# Пример квази-векторизации через np.vecrotize


# Лучше использовать встроенную векторизованную функцию, она в сотни раз быстрее
# res = np.maximum(a, b)

def maxx(x, y):
    "Получите максимум два предмета"
    if x >= y:
        return x
    else:
        return y

a = np.array([5, 7, 9, 8, 6, 4, 5])
b = np.array([6, 3, 4, 8, 9, 7, 1])

res = np.vectorize(maxx, otypes=[float])(a,b)

res
```

```python
# Пример квази-векторизации через np.vecrotize


# Лучше использовать встроенную векторизованную функцию, она в сотни раз быстрее
# res = np.maximum(a, b)

def maxx(x, y):
    "Получите максимум два предмета"
    if x >= y:
        return x
    else:
        return y

a = np.array([5, 7, 9, 8, 6, 4, 5])
b = np.array([6, 3, 4, 8, 9, 7, 1])

res = np.vectorize(maxx, otypes=[float])(a,b)

res
```

```python
# Мин-макс нормализация — это стандартный этап предварительной обработки для алгоритмов машинного обучения 
# (например, KNN, SVM), чувствительных к масштабу признаков.

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
sepallength = np.genfromtxt(url, delimiter=',', dtype='float', usecols=[0])
print(sepallength[:5])

s_max, s_min = sepallength.max(), sepallength.min()
s = (sepallength - s_min) / (s_max - s_min)
print(s)
```

```python

# функция Softmax преобразует исходные оценки в вероятности и является конечной функцией активации в большинстве классификационных нейронных сетей.
# np.exp — векторизованная функция, применяется к каждому элементу массива

# Что здесь делает exp:
# Превращает любые числа (отрицательные, нули, положительные) в положительные
# Усиливает разницу между значениями (большие числа становятся ещё больше)
# После деления на сумму получаем вероятности


url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
sepallength = np.genfromtxt(url, delimiter=',', dtype='float', usecols=[0])

def softmax(x):
    """Численно стабильный softmax."""
    # Вычитаем максимум — это не меняет результат, но предотвращает overflow
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

result = softmax(sepallength)
print("Softmax для sepallength:")
print(result)
print(f"\nСумма: {result.sum():.10f}")  # 1.0000000000
print(f"Мин: {result.min():.6f}")
print(f"Макс: {result.max():.6f}")
```

```python
from scipy.stats import pearsonr

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
iris_2d = np.genfromtxt(url, delimiter=',', dtype='float', usecols=[0,1,2,3])

sepallength=iris_2d[:,0]
petallength=iris_2d[:,2]

pearson=np.corrcoef(sepallength, petallength)[0,1]
print(f'Через numpy: {pearson}')

r, p_value = pearsonr(sepallength, petallength)
print(f'Через scipy:\nКорреляция: {r}\nP_value: {p_value}')

```

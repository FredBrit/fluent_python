```python
# reduce применяет фукнцию к каждому элементу итерируемого объекта и накапливает результат

from functools import reduce
import operator

reduce(operator.add, [1, 2, 3, 4]), reduce(operator.mul, [1, 2, 3, 4]) 
```

```python
reduce(lambda x, y: x if x > y else y, [1, 5, 3, 9, 2])
```

```python
# функция  itertools.count  возвращает  генератор,  порождающий числа. Без аргументов порождается ряд целых чисел, начиная с 0
# Бесконечный счетчик
import itertools
gen = itertools.count(1, .5)
next(gen),next(gen),next(gen)
```

```python
# itertools.takewhile: она порождает генератор, который потребляет другой генератор и останавливается, когда заданный предикат станет равен False.
gen = itertools.takewhile(lambda n: n < 3, itertools.count(1, .5))
list(gen)
```

```python
# По умолчанию бесконечный цикл по итерируемому объекту
from itertools import cycle, islice

list(islice(cycle('AB'), 6))
```

```python
# Повторяет объект N раз
from itertools import repeat
list(repeat((1,2,3), 3))
```

```python
from itertools import accumulate
list(accumulate([1, 2, 3, 4])) # Считается накопительная сумма
```

```python
list(accumulate([3, 1, 4, 2, 5], max)) # accumulate «запоминает» наибольший встреченный элемент и выдаёт его на каждой позиции
# можете передать любую бинарную функцию: min, operator.mul, лямбду и т.д.
```

```python
from itertools import chain
list(chain('ABC', 'DEF'))  # Объединяет итерируемые объекты
```

```python
from itertools import chain
list(chain.from_iterable(['ABC', 'DEF',['RRR','YYY']]))  # Объединяет вложенные итерируемые объекты
```

```python
from itertools import compress
list(compress('ABCDEF', [1, 0, 1, 0, 1, 1])) # Фильтрация по булевым значениям
```

```python
from itertools import dropwhile
list(dropwhile(lambda x: x < 5, [1, 4, 6, 4, 1, 7, 8])) # Игнорирует до наступления первого True условия
```

```python
from itertools import takewhile
list(takewhile(lambda x: x < 5, [1, 4, 6, 4, 1, 7, 8])) # Игнорирует после наступления первого True условия
```

```python
from itertools import filterfalse
list(filterfalse(lambda x: x < 5, [1, 4, 6, 4, 1, 7, 8])) # Фмльтрует итерируемый объект по False условию
```

```python
from itertools import groupby
words = ['cat', 'hi', 'dog', 'elephant', 'bat', 'ant']

result = [(k, list(g)) for k, g in groupby(words, key=len)] # Группирует подряд идущие элементы по ключу, в данному случае, ключ это длина элемента
# При смене значения ключа появляется новая группировка, одинаковых ключей может быть несколько
result
```

```python
from itertools import islice
list(islice('ABCDEFG', 2, 6)) # Получение среза итератора
```

```python
from itertools import pairwise
list(pairwise('ABCD')) # Создание последовательных пар
```

```python
from itertools import starmap
list(starmap(pow, [(2, 5), (3, 2), (10, 3)])) # Применяет функцию при распаковке аргументов
```

```python
from itertools import tee
r, s = tee([1, 2, 3]) # Создает независимые итераторы
list(r), list(s), r is s
```

```python
from itertools import zip_longest
list(zip_longest('ABCD', 'xy', fillvalue='-')) # Соединяет в пары итерируемые объекты по самой длинной последовательности
```

```python
from itertools import product
list(product('AB', 'xy', repeat=2)) # Декартово произведение с возможностью повторения
```

```python
from itertools import permutations
list(permutations('ABC', 2)) #  генерирует все возможные перестановки заданной длины из элементов последовательности без повторений и в соответствии с порядком элементов в итерируемом объекте
```

```python
from itertools import combinations
list(combinations('ABC', 2)) # генерирует все возможные комбинации заданной длины из элементов последовательности без повторений и в соответствии с порядком элементов в итерируемом объекте
```

```python
from itertools import combinations_with_replacement
list(combinations_with_replacement('ABC', 2)) # генерирует все возможные комбинации заданной длины из элементов последовательности с повторениями и в соответствии с порядком элементов в итерируемом объекте
```

```python
def vowel(c):
    return c.lower() in 'aeiou'

list(filter(vowel, 'Aardvark')) # filter из стандартной библиотеки фильтрует последовательность по какой-то функции
```

```python
list(enumerate('albatroz', 1)) # Порождает парные кортежи индекс-элемент. Второй аргумент-номер первого индекса
```

```python
import operator
list(map(operator.mul, range(11), [2, 4, 8])) # Применяет функция к двум и более итерируемым объектам попарно, пока не кончится один из них
```

```python
sum([1, 2, 3, 4]) # Суммирует элементы в итерируемом объекте
```

```python
max([1, 5, 3, 9, 2]), max(['apple', 'banana'], key=len) # Выводит максимальный элемент, либо максимальное значение функции, которая применяется к элементу
```

```python
min([1, 5, 3, 9, 2]), min(['apple', 'banana'], key=len) # Тоже самое, только минимальное
```

```python
any([False, True, False]), any([False, False, False]) # Возвращает True, если хотя бы один элемент в итерируемом объекте удовлетворяет некому условию
```

```python
all([True, True, True]), all([True, False, True]) # Проверка на истинность всех элементов в коллекции
```

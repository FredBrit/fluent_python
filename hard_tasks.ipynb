```python
# Задача на расчет количества одинаковых наборов элементов из массива 
# (например, скольким друзьям можно раздать одинаковый набор конфет)

from collections import Counter
from math import gcd
from functools import reduce

candies=['a', 'b', 'c', 'a', 'b', 'c', 'c', 'c']

def max_friends(candies: list[str]) -> int:
    if not candies:
        return 0
    
    # Получаем частоты каждой конфеты, например: dict_values([3, 5])
    counts = Counter(candies).values()
    
    # reduce последовательно применяет gcd ко всем элементам:
    # gcd(gcd(counts[0], counts[1]), counts[2])...
    # Поиск наиюольшего общего делителя
    return reduce(gcd, counts)

max_friends(candies)    
```

```python
# Идиоматичное решение задачи Самая длинная последовательность через set() вместо сортировки списка

# Если x - 1 есть в множестве, значит x — это продолжение чьей-то другой последовательности, и нам не нужно начинать считать длину с него (мы посчитаем её, когда дойдем до настоящего начала).
# Если x - 1 нет, мы нашли начало! Теперь мы просто запускаем цикл while, проверяя наличие x + 1, x + 2 и так далее, пока последовательность не прервется.

nums=[100, 4, 200, 1, 3, 2]

def longest_consecutive(nums: list[int]) -> int:
    if not nums:
        return 0
        
    num_set = set(nums)  # O(N) время и память для создания сета
    max_len = 0
    
    for num in num_set:
        # Проверяем, является ли num началом последовательности
        if (num - 1) not in num_set:
            current_num = num
            current_len = 1
            
            # Идем вверх по последовательности, пока числа есть в сете
            # Поиск в set работает за O(1)!
            while (current_num + 1) in num_set:
                current_num += 1
                current_len += 1
                
            # Обновляем максимальную длину
            if current_len > max_len:
                max_len = current_len
                
    return max_len    

longest_consecutive(nums)           
```

```python
# Идиоматичное решение задачи распределение подарочных карт между сотрудниками


expectations=[5, 20, 1000] 
cards=[10, 15, 100]

def max_satisfied_teammates(expectations: list[int], cards: list[int]) -> int:
    # 1. Сортируем оба массива по возрастанию
    expectations.sort()
    cards.sort()

    i = 0  # указатель на текущего члена команды
    j = 0  # указатель на текущую подарочную карту
    count = 0  # счетчик удовлетворенных людей

    # 2. Двигаемся по обоим массивам
    while i < len(expectations) and j < len(cards):
        if cards[j] >= expectations[i]:
            # Карта подходит (соответствует или превышает ожидания)
            count += 1
            i += 1  # Переходим к следующему человеку
            j += 1  # Эта карта использована, переходим к следующей
        else:
            # Карта слишком дешевая для этого человека. 
            # Но она может подойти кому-то с меньшими ожиданиями? 
            # Нет, потому что массив отсортирован, и все следующие люди хотят еще больше.
            # Значит, эта карта бесполезна, просто берем следующую.
            j += 1

    return count

max_satisfied_teammates(expectations, cards)
```

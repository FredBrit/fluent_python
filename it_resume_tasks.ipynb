```python
# Раскладывание факториала числа на простые числа со степенями по формуле Легранжа

def is_prime(p):
    for i in range(2, int(p**0.5) + 1):
        if p % i == 0:
            return False
    return True

def foo(n):
    div_res=[]
    pairs=[]
    p=2
    m=1
    while p<=n:
        if is_prime(p)==True:
            d=divmod(n,p**m) 
            if d[0]>0:
                div_res.append(d[0])
                m+=1        
            else:
                pairs.append((p,sum(div_res)))
                div_res=[]
                p+=1
                m=1
        else:
            p+=1
    res = ' * '.join(f'{k}^{v}' if v > 1 else f'{k}' for k,v in pairs)
    return res


foo(6)        
```

```python
# Фильтрация одного списка(итерируемого объекта) по отобранным индексам другого

def solution(nums):
    s1=[el for el in nums]
    s2=[i for i in range(1,len(nums)+1,1) if len(nums)%i==0]
    s3=[s1[i-1] for i in s2]
    res=sum([el**2 for el in s3])
    return res

solution([1, 0, 1, 0, 1])
```

```python
# Мы создаём один итератор на всю строку и многократно вызываем islice, каждый раз забирая из него следующую порцию. 
# Итератор запоминает свою позицию, поэтому следующие вызовы продолжают с того места, где остановились предыдущие.

# O(1) памяти → генератор не хранит все куски сразу, отдаёт их по одному.
# Универсальность → работает не только со строками, но и с файлами, сетевыми потоками, генераторами чисел, где len() недоступен.
# Безопасность остатка → islice корректно обрабатывает последний кусок любой длины (не падает, не теряет данные).
# Скорость → islice написан на C и работает быстрее, чем ручные срезы в цикле.



from itertools import islice
# itertools.islice возвращает итератор, который выдаёт элементы из переданного итерируемого объекта.

s='ABCDEFGHIJKLIMNОQRSTUVWXYZ'
width=4

def chunk_iter(s, width):
    it = iter(s)
    # Превращает строку в итератор. Обычная строка не хранит "текущую позицию". 
    # Итератор же — это объект-указатель, который знает, какой символ будет следующим.
    # Именно благодаря этому мы можем многократно вызывать islice(it, size), 
    # и каждый раз получать следующие символы, а не начинать сначала.
    while True:
        chunk = ''.join(islice(it, width))
        # Важнейшее свойство: islice потребляет элементы из исходного итератора.
        # Когда вы вызываете islice(it, width), она:
        # Забирает из it первые width доступных элемента.
        # Возвращает их как новый итератор.
        # Сдвигает позицию it вперёд. Следующий вызов islice(it, width) возьмёт уже 5-й, 6-й, 7-й и 8-й элементы.
        if not chunk:  # пустая строка = конец
            break
        yield chunk
        # При первом yield функция "засыпает", отдавая кусок наружу. 
        # При следующем обращении к генератору выполнение продолжается со строки chunk = ''.join(...).

res='\n'.join(part for part in chunk_iter(s,width))        
print(res)
```

```python
# Поиск серединного индекса и условие в return

def find_median(s):
    # 1. Сначала в числа, потом сортировка
    nums = sorted(float(x) for x in s.split())
    n = len(nums)
    
    if n % 2 == 0:
        median = (nums[n // 2 - 1] + nums[n // 2]) / 2 # идиоматичный поиск серединного индекса
    else:
        median = nums[n // 2]
        
    # Возвращаем int, если дробная часть нулевая (например, 8.0 -> 8)
    return int(median) if median.is_integer() else round(median,1)
    
find_median('5 2 9 287 7 8 0 156')    
```

```python
# Классический пример на FizzBuzz на match_case

def fizzbuzz(n):
    arr=[i for i in range(1,n+1,1)]
    res=[]
    for el in arr:
        match el:
            case el if el%3==0 and el%5==0: # Строгие условия обязательно должны идти в начале
                res.append('FizzBuzz') 
            case el if el%3==0:
                res.append('Fizz')
            case el if el%5==0:
                res.append('Buzz')   
            case el if el%3!=0:
                res.append(str(el))
    return res

fizzbuzz(15)
```

```python
# Классический счетчик с дефолтными аргументами методов

class Counter:

    def __init__(self):
        self._value=0
    
    def getValue(self):
        return self._value
    
    def increase(self, amount:int=1):
        self._value=self._value+amount
        return amount
    
    def decrease(self, amount:int=1):
        self._value=self._value-amount
        return amount
```

```python
# НЕИДИОМАТИЧНО!!!
# Проверка на равенство произведений двух половин целого числа

from functools import reduce
def foo(n):
    lst=list(str(n))
    middle=int(len(lst)/2)
    for el in lst:
        if len(lst)%2==0:
            right=[int(el) for el in lst[0:middle]]
            left=[int(el) for el in lst[middle:]]
            if reduce(lambda x,y:x*y,right)==reduce(lambda x,y:x*y, left):
                return True
            else:
                return False
        else:
            return False  

foo(n=6426263)              
```

```python
# ИДИОМАТИЧНО!!!
# Проверка на равенство произведений двух половин целого числа


from functools import reduce
from operator import mul

def is_lucky_ticket(number: int) -> bool:
    s = str(number)
    if len(s) % 2 != 0:
        return False
        
    mid = len(s) // 2 # Сразу целочисленное деление
    left = map(int, s[:mid]) # объект-итератор, экономит память
    right = map(int, s[mid:]) # # объект-итератор, экономит память
    
    return reduce(mul, left, 1) == reduce(mul, right, 1) # Для обхода ошибки при пустом объекте TypeError: reduce() of empty sequence with no initial value
```

```python
# Вычисление разницы соседних элементов в 3.10

from itertools import pairwise

def score_of_string(s):
    ords=[ord(el) for el in s]
    res=sum([abs(x-y) for x,y in pairwise(ords)])
    return res

score_of_string('hello')
```

```python
# Вычисление разницы соседних элементов в любой версии

def pairwise_diff(s):
    ords=[ord(el) for el in s]
    it = iter(ords)
    try:
        prev = next(it)
    except StopIteration:
        return  # пустая последовательность
    for curr in it:
        yield abs(curr - prev)
        prev = curr

result = sum(list(pairwise_diff('hello')))
result        
```

```python
# Изменение списка на месте без создания копии

# В Python существует железное соглашение:
# Функции/методы, меняющие объект на месте, должны возвращать None.
# Это прямо зафиксировано в стандартной библиотеке:

# a = [3, 1, 2]
# print(a.sort())        # None
# print(a.reverse())     # None
# print(a.append(4))     # None

# Такой дизайн запрещает цепочки вызовов (arr.move_zeroes().sort()), что защищает от случайной модификации и чётко разделяет:
# in-place → меняет состояние, возвращает None
# functional → возвращает новый объект, оригинал нетронут

arr = [0, 1, 0, 3, 12]
def move_zeroes(arr):
    

    insert_pos = 0
    for i in range(len(arr)):
        if arr[i] != 0:
            arr[insert_pos] = arr[i]
            insert_pos += 1

    # insert_pos уже равен 3 и мы заполняем оставшиеся индексы нулями явно
    for i in range(insert_pos, len(arr)):
        arr[i] = 0
    # Функция ничего не возвращает, можно явно добавить return None
    # И то, и то будет считаться идиоматичным

move_zeroes(arr)
print(arr)       
```

```python
# Пример подсчета количеств подряд идущих элементов и вычисления максимальной подстроки
# Длинный джуновский вариант

from itertools import groupby

s='aaaabbcaaa'
def longest_run(s):
    if not s: # идиоматичная проверка на пустоту
        return 0
    else:
        gr=[(k,len(list(g))) for k,g in groupby(s)]
        d = {}
        for k, v in gr:
            if k not in d or v > d[k]:
                d[k] = v
        res=max(d.values())
        return res

print(longest_run(s))
```

```python
# Классическое решение через рекурсию по алгоритму чисел Фибоначчи
# Сложность алгоритма O(n)

def climb_stairs(n: int) -> int:
    a, b = 1, 1
    for _ in range(n):
        a, b = b, a + b
        print(a,b)
    return a

climb_stairs(5)
```

```python
# Решение через @lru_cache
# Сложность алгоритма O(n)

from functools import lru_cache

@lru_cache(maxsize=None)
def climb_stairs2(n: int) -> int:
    if n <= 1:
        return 1
    return climb_stairs2(n - 1) + climb_stairs2(n - 2) # «дай мне количество способов до предыдущей ступеньки, дай количество до предпредыдущей, сложи их»

climb_stairs2(5)
```

```python
# Генерация непосредственно последовательностей, сложность алгоритма O(2ⁿ)

def generate_ways(n):
    if n == 0:
        yield []          # 0 ступенек → 1 способ: пустой список
    elif n == 1:
        yield [1]         # 1 ступенька → только шаг 1
    else:
        # Берём все способы до n-1 и добавляем в конец шаг 1
        for way in generate_ways(n - 1):
            yield way + [1]
        # Берём все способы до n-2 и добавляем в конец шаг 2
        for way in generate_ways(n - 2):
            yield way + [2]

# Проверка
print(list(generate_ways(3)))
```

```python
# Идиоматичный пример алгоритма бинарного поиска


nums = sorted([1, 3, 5, 7, 9, 2])
target = 9


def search(nums,target):
    low, high = 0, len(nums) - 1
    counter = 0
    pos=-1 # Вернет значение по умолчанию, если список пуст

    while low <= high:
        counter += 1
        mid = (low + high) // 2
        
        if nums[mid] == target:
            pos=mid # Вычисление индекса элемента (mid вычисляется на каждой итерации, так как low и high сдвигаются)
            break  # Нашли!
        elif nums[mid] < target:
            low = mid + 1  # Ищем в правой половине
            
        else:
            high = mid - 1 # Ищем в левой половине

    return counter,pos        

search(nums,target)            
```

```python
# Классическая задача по поиску пары элементов, которые в сумме дают target
# Сложность O(n) по памяти
# Сложность n по времени

nums=sorted([1, 3, 4, 5, 7, 11])
target=14

def two_sum(nums, target):
    seen = {}  # {число: его_индекс}
    for i, num in enumerate(nums):
        complement = target - num
        print(f'complement:{complement},num:{num},i:{i}')
        if complement in seen:
            return [seen[complement], i]  # индекс пары, записанной ранее в seen и индекс только что найденной пары в списке
        seen[num] = i
    return None

two_sum(nums,target)             
```

```python
# Сложность O(1) по памяти
# Сложность n по времени


nums=sorted([1, 3, 4, 5, 7, 11])
target=9

def find_pair_sorted(nums, target):
    left, right = 0, len(nums) - 1
    while left < right:
        print(f'left: {left}({nums[left]}), right: {right}({nums[right]})')
        s = nums[left] + nums[right]
        if s == target:
            return [nums[left], nums[right]]
        elif s < target:
            left += 1    # сумма мала → двигаем левый указатель вправо
        else:
            right -= 1   # сумма велика → двигаем правый указатель влево
    return False

find_pair_sorted(nums, target)
```

```python
# Пример работы с данными в формате JSON

import json
from datetime import datetime,timedelta

ads='[{"id": 1, "date": "2024-03-15"}, {"id": 2, "date": "2024-04-01"}, {"id": 3, "date": "2024-02-28"}]'
threshold='2024-03-20'

def find_outdated(ads, threshold):
    target_date=datetime.strptime(threshold,'%Y-%m-%d').date()
    res=[]
    d=json.loads(ads)
    for el in d:
        for k,v in el.items():
            if k=='date' and datetime.strptime(v,'%Y-%m-%d').date()<target_date:
                res.append(el['id'])
    return res

find_outdated(ads,threshold)            

    
```

```python
# Алгоритм Луна
# Проверка номера банковской карты

number='4532015112830366'

def is_valid_card(number):
    if not number:
        return False
    r_number=list(reversed([int(el) for el in number]))
    doubles=[el*2 if i%2!=0 else el for i,el in enumerate(r_number)]
    validate_doubles=[sum(list(int(num) for num in str(el))) if el>=10 else el for el in doubles]
    return True if sum(validate_doubles)%10==0 else False

is_valid_card(number)
```

```python
# Классическая задача Best Time to Buy and Sell Stock
# Идиоматичное решение

prices=[2, 4, 1]

def max_profit(prices):
    min_price = float('inf') # это встроенная константа Python, обозначающая положительную бесконечность. Строго больше любого реального числа
    max_profit = 0
    for p in prices:
        if p < min_price:
            min_price = p
        elif p - min_price > max_profit:
            max_profit = p - min_price # на последнем элементе max_profit не обновляется, так p - min_price меньше max_profit, которая уже была ранее
    return max_profit

max_profit(prices)
```

```python
# Раскладывание факториала числа на простые числа со степенями по формуле Легранжа

def is_prime(p):
    for i in range(2, int(p**0.5) + 1):
        if p % i == 0:
            return False
    return True

def foo(n):
    div_res=[]
    pairs=[]
    p=2
    m=1
    while p<=n:
        if is_prime(p)==True:
            d=divmod(n,p**m) 
            if d[0]>0:
                div_res.append(d[0])
                m+=1        
            else:
                pairs.append((p,sum(div_res)))
                div_res=[]
                p+=1
                m=1
        else:
            p+=1
    res = ' * '.join(f'{k}^{v}' if v > 1 else f'{k}' for k,v in pairs)
    return res


foo(6)        
```

```python
# Фильтрация одного списка(итерируемого объекта) по отобранным индексам другого

def solution(nums):
    s1=[el for el in nums]
    s2=[i for i in range(1,len(nums)+1,1) if len(nums)%i==0]
    s3=[s1[i-1] for i in s2]
    res=sum([el**2 for el in s3])
    return res

solution([1, 0, 1, 0, 1])
```

```python
# Мы создаём один итератор на всю строку и многократно вызываем islice, каждый раз забирая из него следующую порцию. 
# Итератор запоминает свою позицию, поэтому следующие вызовы продолжают с того места, где остановились предыдущие.

# O(1) памяти → генератор не хранит все куски сразу, отдаёт их по одному.
# Универсальность → работает не только со строками, но и с файлами, сетевыми потоками, генераторами чисел, где len() недоступен.
# Безопасность остатка → islice корректно обрабатывает последний кусок любой длины (не падает, не теряет данные).
# Скорость → islice написан на C и работает быстрее, чем ручные срезы в цикле.



from itertools import islice
# itertools.islice возвращает итератор, который выдаёт элементы из переданного итерируемого объекта.

s='ABCDEFGHIJKLIMNОQRSTUVWXYZ'
width=4

def chunk_iter(s, width):
    it = iter(s)
    # Превращает строку в итератор. Обычная строка не хранит "текущую позицию". 
    # Итератор же — это объект-указатель, который знает, какой символ будет следующим.
    # Именно благодаря этому мы можем многократно вызывать islice(it, size), 
    # и каждый раз получать следующие символы, а не начинать сначала.
    while True:
        chunk = ''.join(islice(it, width))
        # Важнейшее свойство: islice потребляет элементы из исходного итератора.
        # Когда вы вызываете islice(it, width), она:
        # Забирает из it первые width доступных элемента.
        # Возвращает их как новый итератор.
        # Сдвигает позицию it вперёд. Следующий вызов islice(it, width) возьмёт уже 5-й, 6-й, 7-й и 8-й элементы.
        if not chunk:  # пустая строка = конец
            break
        yield chunk
        # При первом yield функция "засыпает", отдавая кусок наружу. 
        # При следующем обращении к генератору выполнение продолжается со строки chunk = ''.join(...).

res='\n'.join(part for part in chunk_iter(s,width))        
print(res)
```

```python
# Поиск серединного индекса и условие в return

def find_median(s):
    # 1. Сначала в числа, потом сортировка
    nums = sorted(float(x) for x in s.split())
    n = len(nums)
    
    if n % 2 == 0:
        median = (nums[n // 2 - 1] + nums[n // 2]) / 2 # идиоматичный поиск серединного индекса
    else:
        median = nums[n // 2]
        
    # Возвращаем int, если дробная часть нулевая (например, 8.0 -> 8)
    return int(median) if median.is_integer() else round(median,1)
    
find_median('5 2 9 287 7 8 0 156')    
```

```python
# Классический пример на FizzBuzz на match_case

def fizzbuzz(n):
    arr=[i for i in range(1,n+1,1)]
    res=[]
    for el in arr:
        match el:
            case el if el%3==0 and el%5==0: # Строгие условия обязательно должны идти в начале
                res.append('FizzBuzz') 
            case el if el%3==0:
                res.append('Fizz')
            case el if el%5==0:
                res.append('Buzz')   
            case el if el%3!=0:
                res.append(str(el))
    return res

fizzbuzz(15)
```

```python
# Классический счетчик с дефолтными аргументами методов

class Counter:

    def __init__(self):
        self._value=0
    
    def getValue(self):
        return self._value
    
    def increase(self, amount:int=1):
        self._value=self._value+amount
        return amount
    
    def decrease(self, amount:int=1):
        self._value=self._value-amount
        return amount
```

```python
# Проверка всех элементов коллекции на соответствие определенному условию

def is_divisible_by_digits(num):

    nums=[int(el) for el in str(num)]
    check = [True if (el != 0 and num % el == 0) or el==0 else False for el in nums]
    return all(check)
```

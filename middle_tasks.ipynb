```python
# Идиоматичная проверка email через регулярки

# \w не учитывает дефис, d\ это цифры во всех кодировках, поэтому лучшие использовать [0-9]
# fullmatch строгая полная проверка паттерна на совпадение

import re

s="user-12-3_@example99.co"

PATTERN = re.compile(r'([a-zA-Z0-9-_]+)@([a-zA-Z0-9]+)\.([a-zA-Z]{1,3})')

def is_valid_email(s):
    return True if re.fullmatch(PATTERN,s) else False

is_valid_email(s)
```

```python
# Валидация строк многострочного css файла

css_code = """#BED {
  color: #fFfDf8;
  background-color:#aef;
  font-size: 123px;
  background: -webkit-linear-gradient(top, #f9f9f9, #fff);
}
#Cab {
  background-color: #ABC;
  border: 2px dashed #fff;
}"""



css_code = '\n'.join([css_code])
def css(css_code):
    print(f'Изначальный css:\n{css_code}\n\n')
    match=re.findall(r'[ :(,]\s*(#[0-9a-fA-F]{6}|#[0-9a-fA-F]{3})(?![0-9a-fA-F])',css_code)

    print(f'Валидные строки:')
    for el in match:
        print(el)
    
css(css_code)
```

```python
# Задача на поиск минимальной амплитуды в массиве, при замене трех элементов

arr=[4, -8, -1, 3, 7, 10, 5]

def min_amplitude(arr):
    if len(arr) <= 4: # Если элементов ≤ 4, можно изменить три элемента и получить амплитуду 0.
        return 0
    
    arr.sort() # написан на C, работает за O(n log n) и быстрее любых ручных циклов. Изменяет список на месте
    # O(1) доп. памяти (сортировка in-place). Никаких копий, словарей или генераторов
    return min(
        arr[-1] - arr[3],  # убрали 3 наименьших
        arr[-2] - arr[2],  # убрали 2 наименьших + 1 наибольший
        arr[-3] - arr[1],  # убрали 1 наименьший + 2 наибольших
        arr[-4] - arr[0]   # убрали 3 наибольших
    )

min_amplitude(arr)
```

```python
# Пример методов с исключениями и использованием внутренних методов для взаимодействия с другими классами

# Золотое правило ООП
# Состояние объекта (баланс, владелец, валюта) self.balance, self.owner
# Параметры операции (сумма перевода, комиссия, целевой счёт) Только в параметрах метода. Не писать в self.


class BankAccount:

    def __init__(self,balance:int=0):
        
        if balance<0:
            raise ValueError("Deposit amount must be positive")
        self.balance=balance

    
    def get_balance(self):
        return self.balance
    
    def deposit(self,amount):
        if amount<=0:
            raise ValueError("Deposit amount must be positive")
        self.balance+=amount
        return self.balance

    def withdraw(self,amount:int):
        if amount<=0:
            raise ValueError("Withdrawal amount must be positive")
        elif self.get_balance()<amount:
            raise ValueError("Insufficient funds")
        self.balance-=amount
        return self.balance
    
    def transfer(self,target_account,amount:int):
        if not isinstance(target_account, BankAccount): # Идиоматичная проверка принадлежности экземпляра к конкретному классу
            raise TypeError("Target must be a BankAccount")
        self.withdraw(amount)
        target_account.deposit(amount)
        return self.balance
```

```python
# Треугольник Паскаля, идиоматичное решение

def pascal_triangle(n):
    res = [[1]]
    for i in range(1, n):
        prev = res[-1]
        # Единицы по краям + суммы соседних элементов внутри
        new_row = [1] + [prev[j] + prev[j+1] for j in range(len(prev)-1)] + [1]
        
        res.append(new_row)
    return res

pascal_triangle(5)
```

```python
# Треугольник Паскаля, решение через вложенный цикл, чтобы понять принцип работы

def pascal_triangle_cycle(n):
    res = [[1]]
    for i in range(1, n):
        prev = res[-1] # Каждый раз обращаемся к последней/следующей строке, то же самое что res[i-1]
        new_row = []
        
        for j in range(len(prev) + 1): # Чтобы избежать Index Error
            print(f'Текущий j: {j}')
            if j == 0 or j == len(prev):
                
                new_row.append(1)  # края
            else:
                
                new_row.append(prev[j - 1] + prev[j])  # внутренние суммы
        print(f'Добавлено: {new_row}')
        res.append(new_row)
    return res

pascal_triangle_cycle(5)

```

```python
# Решение через pairwise

from itertools import pairwise

def pascal_triangle_pairwise(n):
    res = [[1]]
    for i in range(1, n):
        iter_ex=[1, *(a + b for a, b in pairwise(res[i-1])), 1] # на каждой итерации считает сумму соседних элементов
        print(iter_ex)
        res.append(iter_ex)  # внутренние суммы
    return res

pascal_triangle_pairwise(6)
```

```python
# Из-за insert сложность алгоритма O(N**2)
# Лишние затраты памяти на создание списка

def can_place_video_ads(feed_items, n):
    
    indexes = [
    i for i in range(len(feed_items))
    if feed_items[i] == 0
    and (
        (i == 0)      # левый сосед или начало
    or (i == len(feed_items)-1)
     
    or (feed_items[i+1] == 0)
       # правый сосед или конец
        )
    ]
    
    diff=0
    for i in range(len(indexes)):
        if i==0:
            diff+=1
            feed_items.insert(0,1)
        elif i==len(indexes)-1:
            diff+=1
            feed_items.append(1)
        else:
            diff+=1
            feed_items.insert(indexes[i]+diff,1)

    print(feed_items,diff)
    return True if diff==n else False   

can_place_video_ads(feed_items,n) 
```

```python
# Идиоматичное решение задачи про Римские цифры

# O(1) памяти: Мы не создаём никаких списков, только две переменные (total, max_val)
# O(N) времени: Строгий один проход по строке на уровне C (внутри reversed).
# Накопление счетчика total на каждой итерации с вычитанием значения слева(val), если оно больше значения справа(max-val)
# max_val после каждой итерации обновляется, для сравнения соседних элементов

# Словарь вынесен за функцию: он создаётся 1 раз при запуске, а не при каждом вызове

ROMAN_VALUES = {
    'I': 1, 'V': 5, 'X': 10, 'L': 50,
    'C': 100, 'D': 500, 'M': 1000
}

s="MMCDLXIXVIIII"

def roman_to_int(s: str) -> int:
    total = 0
    max_val = 0
    
    # reversed(s) идёт справа налево без создания копий строки
    for char in reversed(s):
        val = ROMAN_VALUES[char]
        if val < max_val:
            total -= val  # Например, I перед V или X
            print(val, max_val, total)
        else:
            total += val
            max_val = val # Обновляем максимум для следующих итераций
            print(val, max_val, total)
    return total


roman_to_int(s)    
```

```python
# Идиоматичное решение задачи по переводу числа в 13-теричную систему

n=1000

def num_13(n):
    # Обработка граничного случая (если вдруг придет 0)
    if n == 0:
        return "0"
    
    # Строка-словарь для маппинга остатков в символы
    # Индекс строки соответствует значению остатка (0-12)
    digits = "0123456789ABC"
    
    res = []
    
    # Пока число больше 0, делим его на 13
    while n > 0:
        remainder = n % 13
        res.append(digits[remainder])
        n = n // 13
        print(n, remainder, digits[remainder])
        
    # Остатки собирались от младшего разряда к старшему, 
    # поэтому переворачиваем список и объединяем в строку
    return "".join(reversed(res))


num_13(n)
```

```python
# Идиоматичная проверка количества слов в строке после очистки

# string.punctuation содержит все символы пунктуации
# maketrans создает таблицу сопоставлений
# sort(key=lambda x: (-x[1], x[0])) идиоматичная сортировка списка на месте по нескольким ключам


import string
from collections import Counter

text="The quick, brown am am, ,dog,  fox jumps over the lazy dog. The dog was not amused."
n=3

def word_frequency(text,n):

    # Создаем таблицу трансляции, где каждому знаку пунктуации сопоставляется None (удалить)
    translator = str.maketrans('', '', string.punctuation)
    text = ' '.join(text.translate(translator).lower().split())

    res=Counter(sorted(text.split()))

    t=[(k,v) for k,v in res.items()]
    t.sort(key=lambda x: (-x[1], x[0]))
    return t[:n]

word_frequency(text,n)




```

```python
# Идиоматичное подсчета количества элементов в массиве и вывод топа

# most_common(k) избавляет от необходимости создания дополнительных списков и выводит топ кортежей

nums=[2, 1, 2, 3, 2, 1, 2, 1, 3, 4]
k=3

from collections import Counter

def top_k_frequent(nums,k):

    res=[a for a,b in Counter(nums).most_common(k)]
    res.sort()
    return res

top_k_frequent(nums,k)
```

```python
# Идиоматичное проверка элементов массива на анаграммность, а также сортировка по нескольким условиям
# Благодаря defaultdict мы проходим массив один раз и затраты по памяти минимальны


# defaultdict в качестве ключей хранит отсортированные буквы каждого слова
# а в качестве значений хранит список слов, которые соответствуют сигнатуре ключа

# Сортировать(изменять) на месте можно не только список, но и каждый его элемент


from collections import defaultdict

words=["a", "on", "cat", "act", "car", "tone", "mute","z", "note"]

def group_anagrams(words):
    # 1. Создаем словарь, где ключ - "подпись" анаграммы, значение - список слов
    anagram_map = defaultdict(list)
    
    for word in words:
        # Сортируем буквы слова, чтобы 'cat' и 'act' дали одинаковый ключ ('a', 'c', 't')
        signature = tuple(sorted(word))
        anagram_map[signature].append(word)
        print(word, signature)
        
    
    # 2. Получаем список всех групп (значения словаря)
    result = list(anagram_map.values())
    print(result)
    
    # 3. Сортируем слова внутри каждой группы по алфавиту
    for group in result:
        group.sort()

    print(result)    
        
    # 4. Сортируем сами группы: сначала по длине слов, потом по алфавиту (берем 1-й элемент группы)
    result.sort(key=lambda x: (len(x[0]), x[0]))
    
    return result


group_anagrams(words)
```

```python
# Неидиоматичный пример нахождения наиболее часто встрачающегося элемента
# Но интересный пример работы конструкции match case

from collections import Counter
cities=["New York, USA", "London, UK", "New York, USA", "Paris, France", "London, UK"]

def most_popular_city(cities):
    
    c=Counter(cities)
    lst=[(k,v) for k,v in c.items()]
    lst.sort(key=lambda x: (-x[1],x[0]) )
    
    city=''
    for el in lst[0:1]:
        match el:
            case (full_name, count):
                city = full_name.split(',')[0].strip()
            case _:
                city = 'Неверный формат'

    return city


most_popular_city(cities)
```

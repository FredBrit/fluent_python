```python
# Задача на поиск минимальной амплитуды в массиве, при замене трех элементов

arr=[4, -8, -1, 3, 7, 10, 5]

def min_amplitude(arr):
    if len(arr) <= 4: # Если элементов ≤ 4, можно изменить три элемента и получить амплитуду 0.
        return 0
    
    arr.sort() # написан на C, работает за O(n log n) и быстрее любых ручных циклов. Изменяет список на месте
    # O(1) доп. памяти (сортировка in-place). Никаких копий, словарей или генераторов
    return min(
        arr[-1] - arr[3],  # убрали 3 наименьших
        arr[-2] - arr[2],  # убрали 2 наименьших + 1 наибольший
        arr[-3] - arr[1],  # убрали 1 наименьший + 2 наибольших
        arr[-4] - arr[0]   # убрали 3 наибольших
    )

min_amplitude(arr)
```

```python
# Пример методов с исключениями и использованием внутренних методов для взаимодействия с другими классами

# Золотое правило ООП
# Состояние объекта (баланс, владелец, валюта) self.balance, self.owner
# Параметры операции (сумма перевода, комиссия, целевой счёт) Только в параметрах метода. Не писать в self.


class BankAccount:

    def __init__(self,balance:int=0):
        
        if balance<0:
            raise ValueError("Deposit amount must be positive")
        self.balance=balance

    
    def get_balance(self):
        return self.balance
    
    def deposit(self,amount):
        if amount<=0:
            raise ValueError("Deposit amount must be positive")
        self.balance+=amount
        return self.balance

    def withdraw(self,amount:int):
        if amount<=0:
            raise ValueError("Withdrawal amount must be positive")
        elif self.get_balance()<amount:
            raise ValueError("Insufficient funds")
        self.balance-=amount
        return self.balance
    
    def transfer(self,target_account,amount:int):
        if not isinstance(target_account, BankAccount): # Идиоматичная проверка принадлежности экземпляра к конкретному классу
            raise TypeError("Target must be a BankAccount")
        self.withdraw(amount)
        target_account.deposit(amount)
        return self.balance
```

```python
# Треугольник Паскаля, идиоматичное решение

def pascal_triangle(n):
    res = [[1]]
    for i in range(1, n):
        prev = res[-1]
        # Единицы по краям + суммы соседних элементов внутри
        new_row = [1] + [prev[j] + prev[j+1] for j in range(len(prev)-1)] + [1]
        
        res.append(new_row)
    return res

pascal_triangle(5)
```

```python
# Треугольник Паскаля, решение через вложенный цикл, чтобы понять принцип работы

def pascal_triangle_cycle(n):
    res = [[1]]
    for i in range(1, n):
        prev = res[-1] # Каждый раз обращаемся к последней/следующей строке, то же самое что res[i-1]
        new_row = []
        
        for j in range(len(prev) + 1): # Чтобы избежать Index Error
            print(f'Текущий j: {j}')
            if j == 0 or j == len(prev):
                
                new_row.append(1)  # края
            else:
                
                new_row.append(prev[j - 1] + prev[j])  # внутренние суммы
        print(f'Добавлено: {new_row}')
        res.append(new_row)
    return res

pascal_triangle_cycle(5)

```

```python
# Решение через pairwise

from itertools import pairwise

def pascal_triangle_pairwise(n):
    res = [[1]]
    for i in range(1, n):
        iter_ex=[1, *(a + b for a, b in pairwise(res[i-1])), 1] # на каждой итерации считает сумму соседних элементов
        print(iter_ex)
        res.append(iter_ex)  # внутренние суммы
    return res

pascal_triangle_pairwise(6)
```

```python
# Идиоматичное решение задачи про Римские цифры

# O(1) памяти: Мы не создаём никаких списков, только две переменные (total, max_val)
# O(N) времени: Строгий один проход по строке на уровне C (внутри reversed).
# Накопление счетчика total на каждой итерации с вычитанием значения слева(val), если оно больше значения справа(max-val)
# max_val после каждой итерации обновляется, для сравнения соседних элементов

# Словарь вынесен за функцию: он создаётся 1 раз при запуске, а не при каждом вызове

ROMAN_VALUES = {
    'I': 1, 'V': 5, 'X': 10, 'L': 50,
    'C': 100, 'D': 500, 'M': 1000
}

s="MMCDLXIXVIIII"

def roman_to_int(s: str) -> int:
    total = 0
    max_val = 0
    
    # reversed(s) идёт справа налево без создания копий строки
    for char in reversed(s):
        val = ROMAN_VALUES[char]
        if val < max_val:
            total -= val  # Например, I перед V или X
            print(val, max_val, total)
        else:
            total += val
            max_val = val # Обновляем максимум для следующих итераций
            print(val, max_val, total)
    return total


roman_to_int(s)    
```

```python
# Идиоматичное решение задачи по переводу числа в 13-теричную систему

n=1000

def num_13(n):
    # Обработка граничного случая (если вдруг придет 0)
    if n == 0:
        return "0"
    
    # Строка-словарь для маппинга остатков в символы
    # Индекс строки соответствует значению остатка (0-12)
    digits = "0123456789ABC"
    
    res = []
    
    # Пока число больше 0, делим его на 13
    while n > 0:
        remainder = n % 13
        res.append(digits[remainder])
        n = n // 13
        print(n, remainder, digits[remainder])
        
    # Остатки собирались от младшего разряда к старшему, 
    # поэтому переворачиваем список и объединяем в строку
    return "".join(reversed(res))


num_13(n)
```

```python
# Идиоматичная проверка количества слов в строке после очистки

# string.punctuation содержит все символы пунктуации
# maketrans создает таблицу сопоставлений
# sort(key=lambda x: (-x[1], x[0])) идиоматичная сортировка списка на месте по нескольким ключам


import string
from collections import Counter

text="The quick, brown am am, ,dog,  fox jumps over the lazy dog. The dog was not amused."
n=3

def word_frequency(text,n):

    # Создаем таблицу трансляции, где каждому знаку пунктуации сопоставляется None (удалить)
    translator = str.maketrans('', '', string.punctuation)
    text = ' '.join(text.translate(translator).lower().split())

    res=Counter(sorted(text.split()))

    t=[(k,v) for k,v in res.items()]
    t.sort(key=lambda x: (-x[1], x[0]))
    return t[:n]

word_frequency(text,n)




```

```python
# Идиоматичное подсчета количества элементов в массиве и вывод топа

# most_common(k) избавляет от необходимости создания дополнительных списков и выводит топ кортежей

nums=[2, 1, 2, 3, 2, 1, 2, 1, 3, 4]
k=3

from collections import Counter

def top_k_frequent(nums,k):

    res=[a for a,b in Counter(nums).most_common(k)]
    res.sort()
    return res

top_k_frequent(nums,k)
```

```python
# Идиоматичное проверка элементов массива на анаграммность, а также сортировка по нескольким условиям
# Благодаря defaultdict мы проходим массив один раз и затраты по памяти минимальны


# defaultdict в качестве ключей хранит отсортированные буквы каждого слова
# а в качестве значений хранит список слов, которые соответствуют сигнатуре ключа

# Сортировать(изменять) на месте можно не только список, но и каждый его элемент


from collections import defaultdict

words=["a", "on", "cat", "act", "car", "tone", "mute","z", "note"]

def group_anagrams(words):
    # 1. Создаем словарь, где ключ - "подпись" анаграммы, значение - список слов
    anagram_map = defaultdict(list)
    
    for word in words:
        # Сортируем буквы слова, чтобы 'cat' и 'act' дали одинаковый ключ ('a', 'c', 't')
        signature = tuple(sorted(word))
        anagram_map[signature].append(word)
        print(word, signature)
        
    
    # 2. Получаем список всех групп (значения словаря)
    result = list(anagram_map.values())
    print(result)
    
    # 3. Сортируем слова внутри каждой группы по алфавиту
    for group in result:
        group.sort()

    print(result)    
        
    # 4. Сортируем сами группы: сначала по длине слов, потом по алфавиту (берем 1-й элемент группы)
    result.sort(key=lambda x: (len(x[0]), x[0]))
    
    return result


group_anagrams(words)
```

```python
# Неидиоматичный пример нахождения наиболее часто встрачающегося элемента
# Но интересный пример работы конструкции match case

from collections import Counter
cities=["New York, USA", "London, UK", "New York, USA", "Paris, France", "London, UK"]

def most_popular_city(cities):
    
    c=Counter(cities)
    lst=[(k,v) for k,v in c.items()]
    lst.sort(key=lambda x: (-x[1],x[0]) )
    
    city=''
    for el in lst[0:1]:
        match el:
            case (full_name, count):
                city = full_name.split(',')[0].strip()
            case _:
                city = 'Неверный формат'

    return city


most_popular_city(cities)
```

```python
# Идиоматичное решение алгоритма Кадане


nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

def max_subarray_sum(nums):

    # Инициализируем первым элементом, чтобы корректно работать с массивами из одних отрицательных чисел
    max_sum = nums[0]
    current_sum = nums[0]

    for el in nums[1:]: # Начинаем со второго элемента
        # ГЛАВНАЯ МАГИЯ: 
        # Мы выбираем: начать новый подмассив с текущего элемента ИЛИ продолжить старый?
        current_sum = max(el, current_sum + el)
        
        # Обновляем глобальный максимум
        max_sum = max(max_sum, current_sum)
        
        print(f"Элемент: {el:2} | Текущая сумма: {current_sum:2} | Максимум: {max_sum}")

    print(f"\nИтоговый ответ: {max_sum}")

    return max_sum

max_subarray_sum(nums)

```

```python
# Задач на перекрытие интервалов


intervals=[[1,3],[2,6],[8,10],[15,18]]


def merge(intervals: list[list[int]]) -> list[list[int]]:
    if not intervals:
        return []

    # 1. Сортируем интервалы по времени начала (первому элементу)
    intervals.sort(key=lambda x: x[0])
    
    # 2. Инициализируем результат первым интервалом
    merged = [intervals[0]]
    
    # 3. Проходим по всем остальным интервалам
    for current in intervals[1:]:
        last_merged = merged[-1] # Берем последний добавленный в результат интервал
        
        # Проверяем перекрытие: начало текущего <= конца последнего объединенного
        if current[0] <= last_merged[1]:

            print(current, last_merged, current[0], last_merged[1], merged)

            # Перекрываются! Обновляем конец последнего интервала максимумом из двух
            last_merged[1] = max(last_merged[1], current[1])
        else:
            # Не перекрываются. Добавляем текущий интервал как новый
            print(current, last_merged, current[0], last_merged[1], merged)
            merged.append(current)
            
    return merged

merge(intervals)
```

```python
# Идиоматичное решение задачи поиска количества комбинаций элементов массива в сумме дающих определенное число, а также вывод самой короткой комбинации или количества элементов самой короткой комбинации


def find_all_combinations(coins, target):
    # Убираем дубликаты из списка монет и сортируем их по возрастанию.
    # Сортировка нужна для оптимизации: если текущая монета уже дает перебор, 
    # следующие (более крупные) монеты тем более дадут перебор.
    coins = sorted(list(set(coins)))
    
    all_combinations = []

    # Внутренняя рекурсивная функция
    def backtrack(start_index, current_sum, current_path):
        # Базовый случай 1: мы нашли точную сумму
        if current_sum == target:
            all_combinations.append(current_path[:]) # Добавляем КОПИЮ списка
            return
        
        # Базовый случай 2: мы превысили сумму, дальше идти бессмысленно
        if current_sum > target:
            return

        # Перебираем монеты, начиная с start_index
        for i in range(start_index, len(coins)):
            coin = coins[i]
            
            # Оптимизация: если добавление этой монеты уже дает перебор, 
            # прерываем цикл (так как список отсортирован, следующие монеты еще больше)
            if current_sum + coin > target:
                break
            
            # 1. Делаем выбор: добавляем монету в текущий путь
            current_path.append(coin)
            
            # 2. Рекурсивно вызываем функцию для следующего шага.
            # Передаем 'i' (а не 'i+1'), потому что одну и ту же монету можно использовать много раз.
            backtrack(i, current_sum + coin, current_path)
            
            # 3. Откатываемся назад (Backtrack): удаляем последнюю монету, 
            # чтобы на следующей итерации цикла попробовать другой вариант.
            current_path.pop()

    # Запускаем рекурсию с нулевой суммы, пустого пути и нулевого индекса
    backtrack(0, 0, [])
    
    if all_combinations:
        result=len(min(all_combinations, key=len))
    else:
        result=-1
    return result


# --- Проверка ---
coins = [1, 2, 4]
target = 5

find_all_combinations(coins, target)
```

```python
# Решение задачи на RLE-декодирование через регулярки

import re

s="ab13c2d4e"

import re

def rle_decode(encoded_string: str) -> str:
    # \d* означает "ноль или более цифр"
    pairs = re.findall(r'([a-z])(\d*)', encoded_string)

    print(pairs)
    
    result = []
    for char, count in pairs:
        # Если группа с цифрами пустая, заменяем на "1"
        if count == "":
            count = "1"
        result.append(char * int(count))
        
    return ''.join(result)

rle_decode(s)
```

```python
# Идиоматичное решение без регулярок

# Вложенный цикл while проверяет следующий строковый элемент на то, является ли он на самом деле числом
# Формирует count_str из таких цифр для каждой пары и если у пары нет числа вставляет 1


s="ab13c2d4e"

def rle_decode2(s: str) -> str:
    result = []
    i = 0
    n = len(s)
    
    while i < n:
        char = s[i]
        i += 1
        
        count_str = ""
        # Собираем все цифры подряд
        while i < n and s[i].isdigit():
            count_str += s[i]
            i += 1
            
        # НОВОЕ: Если цифр не было, по умолчанию ставим 1
        if not count_str:
            count_str = "1"
            
        result.append(char * int(count_str))
        
    return ''.join(result)

rle_decode2(s)
```

```python
# Решение задачи на нахождение угла между часовой и минутной стрелкой

# Работает, но для времени 11:59 цикл сделает 719 итераций и может накапливаться погрешность измерений

hour = 3
minutes = 0

def clock_angle(hour, minutes):

    # Переводим всё целевое время в общее количество минут
    total_target_minutes = (hour * 60) + minutes

    tek_hour = 0.0
    tek_minutes = 0.0

    # Делаем ровно столько шагов, сколько минут прошло с 00:00
    for _ in range(total_target_minutes):
        tek_minutes += 6
        tek_hour += 0.5
        
        # Если минутная стрелка сделала круг, сбрасываем её для чистоты (опционально)
        if tek_minutes == 360:
            tek_minutes = 0

    # Вычисляем разницу
    res = abs(tek_hour - tek_minutes)

    # ВАЖНО: Угол на часах не может быть больше 180 градусов.
    # Если разница, например, 350, реальный угол равен 360 - 350 = 10.
    print(f' Часовая: {tek_hour}, минутная: {tek_minutes}')
    return 360-res if res>180 else res

clock_angle(hour, minutes)
```

```python
# Идиоматичное математическое решение


def get_angle(h, m):
    h = h % 12
    angle_h = (h * 30) + (m * 0.5)
    angle_m = m * 6
    diff = abs(angle_h - angle_m)
    return min(diff, 360 - diff)

get_angle(2,30)
```

```python
# Идиоматичная проверка строк на изоморфность с помощью хеш-таблиц и перекрестного добавления элементовс проверкой по ключу


s='egg'
t='adr'

def is_isomorphic(s: str, t: str) -> bool:
    # Если длины разные, они точно не изоморфны
    if len(s) != len(t):
        return False
    
    # Словарь для проверки: символ из s -> символ из t
    map_s_to_t = {}
    # Словарь для проверки: символ из t -> символ из s
    map_t_to_s = {}
    
    # Проходим по обеим строкам одновременно с помощью zip
    for char_s, char_t in zip(s, t):
        
        # ПРОВЕРКА 1: Если мы уже видели char_s, он должен вести к тому же char_t
        if char_s in map_s_to_t:
            if map_s_to_t[char_s] != char_t:
                print(map_s_to_t[char_s], char_t)
                return False
        else:
            map_s_to_t[char_s] = char_t
            print(map_s_to_t[char_s], char_t)
            
        # ПРОВЕРКА 2: Если мы уже видели char_t, он должен исходить из того же char_s
        if char_t in map_t_to_s:
            if map_t_to_s[char_t] != char_s:
                print(map_t_to_s[char_t], char_s)
                return False
        else:
            map_t_to_s[char_t] = char_s
            print(map_t_to_s[char_t], char_s)
            
    # Если цикл прошел до конца и конфликтов не было, строки изоморфны
    return True

is_isomorphic(s,t)
```

```python
# Идиоматичное решение задачи с нахождением последовательности элементов с наибольшим средним

# O(N) по времени, О(1) по памяти


nums = [1, 2, -5, -3, 10, 3]
k = 3

def findMaxAverage(nums: list[int], k: int) -> float:
    # 1. Считаем сумму самого первого окна длины k
    current_sum = sum(nums[:k])
    max_sum = current_sum
    print(nums[:k])
    print(current_sum)

    # 2. "Скользим" окном по остальной части массива
    # i начинает с k-го элемента (первый элемент нового окна)
    for i in range(k, len(nums)):
        # Вычитаем элемент, который остался слева за бортом, 
        # и прибавляем новый элемент, который вошел справа
        current_sum = current_sum - nums[i - k] + nums[i]
        print(nums[i-k], nums[i])
        # Обновляем максимум (сравниваем суммы, а не средние, чтобы не терять точность)
        if current_sum > max_sum:
            max_sum = current_sum
            
    # 3. Округляем только финальный результат
    return round(max_sum / k, 2)


findMaxAverage(nums, k)
```

```python
# Решение задачи по переводу двоичных чисел в десятичные и обратно без питоновских методов, только математика


a='1010' 
b='1011'

def add_binary(a, b):
    if a=="0" or b=="0":
        return "0"
    a_10=0
    b_10=0

    for i, el in enumerate(reversed(a)):
        a_10+=int(el)*(2**i)
    
    for i, el in enumerate(reversed(b)):
        b_10+=int(el)*(2**i)
    
    summ=a_10+b_10
    
    ostatok=[]
    while summ>0:
        ostatok.append(str(summ%2))
        summ//=2
        
    
    res=''.join(reversed(ostatok))
    return res

add_binary(a,b)      
```

```python
# Идиоматичное решение для возврата строки с сохранением порядка слов, но с обратным порядком букв в словах

# Вместо reversed,list для каждого элемента лучше использовать конструкцию word[::-1], это идиоматичный переворот списка и очень быстрый без лишних действий

s='The quick brown fox jumps over the lazy dog'

def reverse_letters(input_string: str) -> str:
    return ' '.join(word[::-1] for word in input_string.split())

reverse_letters(s)
```

```python
# Идиоматичная проверка количества "настоящих" четных чисел

# Игнорируем нечетные индексы и снижаем длину в два раза
# Вместо целочисленных делений и проверки каждой цифры в числе, мы проверяем состоит ли это число из четных цифр '02468'


n=25

def count_honestly_even(n: int) -> int:
    # Идем только по четным числам (шаг 2), начиная с 2
    # all() проверяет, что каждая цифра есть в строке '02468'
    return sum(1 for i in range(2, n + 1, 2) if all(c in '02468' for c in str(i)))

count_honestly_even(n)
```

```python
# Идиоматичное решение задачи по свертыванию элементов массива

numbers=[1, 2, 3, 5, 7, 8, 9, 10]

def summarize_ranges(numbers):
    if not numbers:
        return ""
    
    numbers.sort()
    res = []
    start = numbers[0]  # Запоминаем начало текущей группы
    
    # Начинаем со второго элемента (индекс 1)
    for i in range(1, len(numbers)):
        prev = numbers[i-1]  # Предыдущее число
        curr = numbers[i]    # Текущее число
        
        # ЯВНАЯ ПРОВЕРКА: продолжается ли последовательность?
        if curr == prev + 1:
            # ДА, продолжается. 
            # Мы просто идем дальше. Переменная start НЕ меняется, 
            # потому что группа всё ещё тянется от старого start.
            print(f'start: {start}, curr: {curr}, prev: {prev}')
            pass 
            
        else:
            # НЕТ, произошел РАЗРЫВ (например, было 3, а стало 5).
            # Значит, старая группа закончилась на числе prev.
            
            # 1. Сохраняем завершенную группу в результат
            if start == prev:
                res.append(str(start))
                print(f'start: {start}, curr: {curr}, prev: {prev}')          # Это было одиночное число (например, 5)
            else:
                res.append(f"{start}-{prev}")
                print(f'start: {start}, curr: {curr}, prev: {prev}')   # Это был диапазон (например, 1-3)
            
            # 2. Начинаем новую группу! 
            # Текущее число (curr) становится началом нового диапазона.
            start = curr  

    # ⚠️ ВАЖНО: Цикл закончился, но последняя группа еще не сохранена в res!
    # Сохраняем её сейчас, используя последнее число массива
    last_num = numbers[-1]
    if start == last_num:
        res.append(str(start))
        print(f'start: {start}, last_num: {last_num}')
    else:
        res.append(f"{start}-{last_num}")
        print(f'start: {start}, last_num: {last_num}')

        
    return ",".join(res)


summarize_ranges(numbers)    
```

```python
# Идиоматичное решение задачи на самую длинную паузу между будильниками

alarms=[60, 600, 720]

def longest_silence(alarms):
    if not alarms:
        return 1440
    
    # 1. Сортируем будильники по времени
    alarms.sort()
    
    # 2. Добавляем виртуальные границы:
    # -1: чтобы тишина до первого будильника считалась как alarms[0] - (-1) - 1 = alarms[0]
    # 1440: чтобы тишина после последнего считалась как 1440 - alarms[-1] - 1
    extended = [-1] + alarms + [1440]
    print(extended)
    
    max_silence = 0
    
    # 3. Считаем разрывы между соседними элементами
    for i in range(len(extended) - 1):
        # Формула разрыва: (следующий будильник) - (текущий будильник) - 1
        # Вычитаем 1, потому что будильник занимает минуту [t, t+1)
        silence = extended[i+1] - extended[i] - 1
        
        if silence > max_silence:
            max_silence = silence
            
    return max_silence

longest_silence(alarms)     

```

```python
# Идиоматичное решение задачи
# Найдите длину самой длинной подстроки (то есть непрерывного отрезка символов), которая является палиндромом


s='baBadDdadbBbdaaaa'

def longest_palindrome(s: str) -> int:
    if not s:
        return 0
    
    max_len = 0
    
    # Вспомогательная функция для расширения от центра
    def expand_around_center(left: int, right: int) -> int:
        # Расширяем, пока мы в границах строки И символы совпадают
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
            print(f'left: {left}, right: {right}')
        
        # Когда цикл прервался, valid палиндром был между (left + 1) и (right - 1)
        # Его длина вычисляется как: (right - 1) - (left + 1) + 1 = right - left - 1
        return right - left - 1

    # Перебираем каждый символ как потенциальный центр
    for i in range(len(s)):
        # 1. Центр — это один символ (для нечетных палиндромов, например "aba")
        len_nechet = expand_around_center(i, i)
        
        # 2. Центр — это промежуток между символами (для четных, например "abba")
        len_chet = expand_around_center(i, i + 1)
        
        # Обновляем максимум, выбирая наибольшее из трех значений
        max_len = max(max_len, len_nechet, len_chet)
        print(f'max_len: {max_len}, len_chet: {len_chet}, len_nechet: {len_nechet}')
        
    return max_len

longest_palindrome(s)
```

```python
# Идиоматичное решение задачи на количество банковских купюр образующих сумму

amount=100550

def get_money(amount: int) -> dict:
    if amount <= 0:
        return {}
    
    # Номиналы в порядке убывания (можно использовать int, это правильнее для математики)
    denominations = [5000, 1000, 500, 100, 50]
    result = {}
    
    for denom in denominations: # Всегд 5 итераций
        # Сколько купюр этого номинала помещается в текущую сумму?
        count = amount // denom
        print(f'amount: {amount}, denom: {denom}, count: {count}')
        
        if count > 0:
            # Если автотест строго требует строковые ключи, используйте str(denom)
            result[str(denom)] = count  
            
            # Оставляем только остаток для следующих, более мелких купюр
            amount %= denom 
            
    return result

get_money(amount)
```

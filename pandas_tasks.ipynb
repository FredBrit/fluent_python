```python
# Фильтрация значений в Series с помощью маски

ser1 = pd.Series([1, 2, 3, 4, 5])
ser2 = pd.Series([4, 5, 6, 7, 8])

# Метод isin() проверяет, есть ли каждый элемент ser1 внутри ser2. Он возвращает булеву маску 
# ~ отрицает True условие
res=ser1[~ser1.isin(ser2)]
res
```

```python
# Нахождение симметрической разницы
# Тоже самое, что ниже, но короче 
# pd.Series(np.setxor1d(ser1, ser2))

ser1 = pd.Series([1, 2, 3, 4, 5])
ser2 = pd.Series([4, 5, 6, 7, 8])

# np.union1d(ser1, ser2) - Объединение
# Она берет два массива, объединяет их и убирает дубликаты, возвращая отсортированный массив уникальных значений.
ser_u = pd.Series(np.union1d(ser1, ser2))  # union

# np.intersect1d(ser1, ser2) — Пересечение
# Возвращает только те элементы, которые есть в обоих массивах одновременно (тоже отсортированные и уникальные).
ser_i = pd.Series(np.intersect1d(ser1, ser2))  # intersect
print(ser_i)

# Отрицание маски
ser_u[~ser_u.isin(ser_i)]
```

```python
# Создаем генератор с фиксированным зерном
rng = np.random.default_rng(100)

# Используем его метод integers вместо randint
ser = pd.Series(rng.integers(1, 5, size=12))

# Находим два самых частых значения
top2 = ser.value_counts().head(2).index
print(top2)

# Оставляем их, остальное заменяем на "Другое"
result = ser.where(ser.isin(top2), 'Другое')

# Либо так:
# result = pd.Series(np.where(ser.isin(top2), ser, 'Другое'))

# Либо так:
# result = ser.mask(~ser.isin(top2), 'Другое')

print(result)
```

```python
# Децили (Deciles)
# Определение: Делят данные на 10 равных частей по количеству наблюдений. Каждый дециль содержит ~10% данных.

# Квартили (Quartiles)
# Определение: Делят данные на 4 равные части. Каждый квартиль содержит ~25% данных.

# Процентили (Percentiles)
# Определение: Делят данные на 100 равных частей. Каждый процентиль содержит ~1% данных.
```

```python
# Децили — это квантили, которые делят данные на 10 равных частей по количеству наблюдений. 
# То есть в каждом дециле будет примерно одинаковое число элементов (в нашем случае: 20 / 10 = по 2 элемента в каждом дециле).

ser = pd.Series(np.random.random(20))

# Разбиваем на 10 децилей с именованными метками
# q=10 — на сколько частей разбить (10 децилей)
# Делит на группы с одинаковым количеством элементов
result = pd.qcut(ser, q=10, labels=[f'D{i}' for i in range(1, 11)])

# Делит на группы с одинаковой шириной интервала
result2= pd.cut(ser, bins=10).value_counts()

print(result)
print(result2)
```

```python
# MSE (Mean Squared Error) — это метрика, которая показывает, насколько сильно предсказания отличаются от реальных значений.
# Среднее значение квадратов разностей между истиной и предсказанием 
# Чем меньше MSE, тем точнее предсказания.


truth = pd.Series(range(10))
pred = pd.Series(range(10)) + np.random.random(10)

# Способ 1: вручную (по формуле)
mse = ((truth - pred) ** 2).mean()
print(f"MSE: {mse:.4f}")

# Способ 2: через sklearn (профессиональный)
from sklearn.metrics import mean_squared_error
mse_sklearn = mean_squared_error(truth, pred)
print(f"MSE (sklearn): {mse_sklearn:.4f}")
```

```python
# Преобразование любого формата даты в datetime64
# format='mixed' для каждого элемента свой шаблон
# dayfirst=True Интерпретирует 02-03-2011 как 2 марта (европейский стиль)

ser = pd.Series(['01 Jan 2010', '02-02-2011', '20120303', '2013/04/04', '2014-05-05', '2015-06-06T12:20'])
res=pd.to_datetime(ser, format='mixed', dayfirst=True)
res
```

```python
# После преобразования в datetime, можно использовать методы dt, чтобы выташить и отформатировать дату как захочется

ser = pd.Series(['01 Jan 2010', '02-02-2011', '20120303', '2013/04/04', '2014-05-05', '2015-06-06T12:20'])

datetimes=pd.to_datetime(ser, format='mixed', dayfirst=True)

days=datetimes.dt.day.to_list()
weeks=datetimes.dt.isocalendar().week.to_list()
day_nums=datetimes.dt.day_of_year.to_list()
days_of_week=datetimes.dt.day_name().to_list()

print(days)
print(weeks)
print(day_nums)
print(days_of_week)
```

```python
ser = pd.Series(['Jan 2010', 'Feb 2011', 'Mar 2012'])
datetimes = pd.to_datetime(ser, format='%b %Y')  # убираем предупреждение

# Добавляем к datetime64 какой-то период через pd.to_timedelta
result = datetimes + pd.to_timedelta(3, unit='D')

print(result)
```

```python
# Фильтрация значений в Series с помощью маски

ser1 = pd.Series([1, 2, 3, 4, 5])
ser2 = pd.Series([4, 5, 6, 7, 8])

# Метод isin() проверяет, есть ли каждый элемент ser1 внутри ser2. Он возвращает булеву маску 
# ~ отрицает True условие
res=ser1[~ser1.isin(ser2)]
res
```

```python
# Нахождение симметрической разницы
# Тоже самое, что ниже, но короче 
# pd.Series(np.setxor1d(ser1, ser2))

ser1 = pd.Series([1, 2, 3, 4, 5])
ser2 = pd.Series([4, 5, 6, 7, 8])

# np.union1d(ser1, ser2) - Объединение
# Она берет два массива, объединяет их и убирает дубликаты, возвращая отсортированный массив уникальных значений.
ser_u = pd.Series(np.union1d(ser1, ser2))  # union

# np.intersect1d(ser1, ser2) — Пересечение
# Возвращает только те элементы, которые есть в обоих массивах одновременно (тоже отсортированные и уникальные).
ser_i = pd.Series(np.intersect1d(ser1, ser2))  # intersect
print(ser_i)

# Отрицание маски
ser_u[~ser_u.isin(ser_i)]
```

```python
# Децили — это квантили, которые делят данные на 10 равных частей по количеству наблюдений. 
# То есть в каждом дециле будет примерно одинаковое число элементов (в нашем случае: 20 / 10 = по 2 элемента в каждом дециле).

ser = pd.Series(np.random.random(20))

# Разбиваем на 10 децилей с именованными метками
# q=10 — на сколько частей разбить (10 децилей)
# Делит на группы с одинаковым количеством элементов
result = pd.qcut(ser, q=10, labels=[f'D{i}' for i in range(1, 11)])

# Делит на группы с одинаковой шириной интервала
result2= pd.cut(ser, bins=10).value_counts()

print(result)
print(result2)
```

```python
# MSE (Mean Squared Error) — это метрика, которая показывает, насколько сильно предсказания отличаются от реальных значений.
# Среднее значение квадратов разностей между истиной и предсказанием 
# Чем меньше MSE, тем точнее предсказания.


truth = pd.Series(range(10))
pred = pd.Series(range(10)) + np.random.random(10)

# Способ 1: вручную (по формуле)
mse = ((truth - pred) ** 2).mean()
print(f"MSE: {mse:.4f}")

# Способ 2: через sklearn (профессиональный)
from sklearn.metrics import mean_squared_error
mse_sklearn = mean_squared_error(truth, pred)
print(f"MSE (sklearn): {mse_sklearn:.4f}")
```

```python
# Нахождение пиков в Series

ser = pd.Series([2, 10, 3, 4, 9, 10, 2, 7, 3])

# Сравниваем каждый элемент с соседями
# shift(1) — сдвиг на 1 вправо (это "сосед слева")
# shift(-1) — сдвиг на 1 влево (это "сосед справа")
peak_mask = (ser > ser.shift(1)) & (ser > ser.shift(-1))

# Получаем индексы пиков
peak_indices = ser[peak_mask].index.tolist()
print("Индексы пиков:", peak_indices)


# Получаем значения пиков
peak_values = ser[peak_mask].tolist()
print("Значения пиков:", peak_values)
```

```python
# Автокорреляция — это мера того, насколько значения ряда связаны с его же значениями, но сдвинутыми на k позиций (лаг k).
# Простыми словами: "Насколько сегодняшнее значение похоже на значение k дней назад?"

ser = pd.Series(np.arange(20) + np.random.normal(1, 10, 20))

# Вычисляем автокорреляции для лагов 1-10
# Встроенный метод pandas autocorr, который считает корреляцию Пирсона между рядом и его сдвинутой версией
autocorr = [ser.autocorr(lag=i) for i in range(1, 11)]

for i, corr in enumerate(autocorr, start=1):
    print(f" {i:2d} | {corr:.4f}")

# Находим индекс максимального значения и добавляем 1 (т.к. нумерация лагов с 1)
max_lag = np.argmax(autocorr) + 1 
max_corr = max(autocorr)
print(f"\nНаибольшая автокорреляция: {max_corr:.4f} при лаге {max_lag}")
```

```python
# Универсальная функция обмена столбцов


df = pd.DataFrame(np.arange(20).reshape(-1, 5), columns=list('abcde'))

print(df)


def swap_columns(df, col1, col2):
    """Меняет местами два столбца по их именам."""
    col_list = list(df.columns)
    # Находим позиции столбцов
    idx1, idx2 = col_list.index(col1), col_list.index(col2)
    # Меняем местами
    col_list[idx1], col_list[idx2] = col_list[idx2], col_list[idx1]
    # Возвращаем DataFrame с новым порядком колонок
    return df[col_list]

# Пример использования: меняем 'a' и 'd'
df_swapped = swap_columns(df, 'a', 'd')
print('После замены')
print(df_swapped)
```

```python
# Все опции вывода

# display.max_rows
# Макс. строк при печати

# display.max_columns
# Макс. столбцов при печати

# display.min_rows
# Мин. строк при усечении

# display.width
# Ширина окна (в символах)

# display.max_colwidth
# Макс. ширина содержимого ячейки

# display.precision
# Точность чисел с плавающей точкой

# display.float_format
# Формат вывода float

# Сброс настроек
# pd.reset_option('^display.')

df = pd.DataFrame(np.arange(100).reshape(-1, 4))


# Устанавливаем лимиты
pd.set_option('display.max_rows', 10)       # максимум 10 строк
pd.set_option('display.max_columns', 2)    # максимум 2 столбца

# Проверяем
print(pd.get_option('display.max_rows')) 
print(pd.get_option('display.max_columns'))


print(df)
```
